# Based on the festures we determine whetherthe breast cancer is of Bengin tumor or Malignt tumor

#### Here we are using logistic regression model (binary classification purpose)

#### Importing the dependencies or packages

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

#### Data collection and processing

```python
from sklearn.datasets import load_breast_cancer
breast_cancer_dataset = load_breast_cancer()
```

```python
print(breast_cancer_dataset)
```

```python
# load the data into a pandas DataFrame
data_frame = pd.DataFrame(breast_cancer_dataset.data, columns = breast_cancer_dataset.feature_names)
```

```python
data_frame.head()
```

```python
data_frame.columns
```

```python
data_frame.info()
```

```python
# adding the 'target' column to the data frame
data_frame['label'] = breast_cancer_dataset.target
```

```python
data_frame.tail()
```

```python
data_frame.shape
```

```python
# Checking for missing values
data_frame.isnull().sum()
```

#### Hence there are no missing or null values

```python
# checking the distribution of Target Varibale
data_frame['label'].value_counts()
```

### 1 -> Benign
### 0 -> Malignant

```python
# Independent and Dependent features
X = data_frame.drop(columns='label')
Y = data_frame['label']
```

```python
X
```

```python
Y
```

```python
# train test split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

### Model Training (Logistic Model)

```python
model = LogisticRegression()
model.fit(X_train, Y_train)
```

### Evaluating Model Performance

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
print("Accuracy:",accuracy_score(Y_test, y_pred))
print("Classification Report:\n",classification_report(Y_test, y_pred))
print("Confusion Matrix:\n",confusion_matrix(Y_test, y_pred))
```

### Hyperparamter tuning

1) Using GridSearchCV

```python
penalty = ['l1', 'l2','elasticnet', 'none']
c_values = [0.01, 0.1, 1, 10, 100]
solver = ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']

grid_params = {'penalty': penalty, 'C': c_values, 'solver': solver}
```

```python
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
## GridSearchCV for hyperparameter tuning
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(estimator=model, param_grid=grid_params, cv=cv, scoring='accuracy', n_jobs=-1)
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
grid.fit(X_train, Y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
grid_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
print("Accuracy:",accuracy_score(Y_test, grid_pred))
print("Classification Report:\n",classification_report(Y_test, grid_pred))
print("Confusion Matrix:\n",confusion_matrix(Y_test, grid_pred))
```

2) Using RandomizedCV

```python
from sklearn.model_selection import RandomizedSearchCV
random = RandomizedSearchCV(estimator=model, param_distributions=grid_params, cv=5, scoring='accuracy', n_jobs=-1, n_iter=10, random_state=42)
```

```python
random.fit(X_train, Y_train)
```

```python
random.best_params_
```

```python
random.best_score_
```

```python
random_pred = random.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
print("Accuracy:",accuracy_score(Y_test, random_pred))
print("Classification Report:\n",classification_report(Y_test, random_pred))
print("Confusion Matrix:\n",confusion_matrix(Y_test, random_pred))
```

### Hence 
- Logistic Model :- 92%
- GridSearchCV :- 93%
- RandomizedCV :- 92%

So, GridSearchCV gives more Accuracy.
